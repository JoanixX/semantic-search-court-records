```python
import pandas as pd
from scipy.stats import trim_mean
```

```python
# Cargar las métricas de las 100 corridas
df_secuencial = pd.read_csv('../../cmd/secuential_cleaner/metricas_secuencial.csv')

print("Dataset de resultados secuenciales cargados:")
df_secuencial.head()
```

```python
# Calcular media recortada al 10% (0.1)
media_recortada_sec = trim_mean(df_secuencial['tiempo_segundos'], 0.1)
tiempo_proyectado_1M = (media_recortada_sec / 20000) * 149387 # Ajusta al total de tu dataset

print(f"Media recortada secuencial (20k registros): {media_recortada_sec:.4f} s")
print(f"Proyección total estimada (1.4M registros): {tiempo_proyectado_1M:.4f} s")
```
